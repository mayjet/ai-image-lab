# FLUX.2 [klein] LoRA training

Base 4Bを公式Diffusers trainerで学習します。最初は必ずdry-runでコマンドを確認してください。Jetsonでは学習を開始しません。

```python
from pathlib import Path
import subprocess, sys

PROJECT_ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / '.git').exists())
CHARACTER = 'character-name'  # datasets/works/<character>/lora
TRIGGER_PROMPT = 'ZK_CHAR, one anime girl'
RUN_NAME = 'zk-char-flux2'
RESOLUTION = 512
STEPS = 1000
print(PROJECT_ROOT)
```

```python
base_command = [
    sys.executable, str(PROJECT_ROOT / 'flux2/train_lora.py'),
    '--character', CHARACTER,
    '--instance-prompt', TRIGGER_PROMPT,
    '--run-name', RUN_NAME,
    '--resolution', str(RESOLUTION),
    '--steps', str(STEPS),
]
subprocess.run([*base_command, '--dry-run'], cwd=PROJECT_ROOT, check=True)
```

## 学習開始

24 GB級のx86 NVIDIA GPUでのみ `RUN_TRAIN = True` にします。Notebook kernelにモデルを保持せず、別プロセスで実行するため、終了後にRAM/VRAMが回収されます。

```python
RUN_TRAIN = False
if RUN_TRAIN:
    subprocess.run([*base_command, '--train'], cwd=PROJECT_ROOT, check=True)
else:
    print('RUN_TRAIN=False: 学習は開始していません')
```
